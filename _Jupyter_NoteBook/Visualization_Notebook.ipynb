### 📘 Cosmetics Customer Dataset — Visualization Notebook

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\Mayuresh\OneDrive\Desktop\_Project_\_DataSet_\_Cleaned_Cosmetic_Dataset.csv")
```

### Customer Age Distribution

```python
sns.histplot(df['Age'], kde=True)
plt.title("Customer Age Distribution")
plt.show()
```

### Top 5 Cities by Orders

```python
df.groupby('City')['Total_Orders'].sum().sort_values(ascending=False).head(5).plot(kind='bar')
plt.title("Top 5 Cities by Total Orders")
plt.show()
```

### Preferred Category Share

```python
df['Preferred_Category'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Preferred Category Distribution")
plt.show()
```

### Discount Used vs Orders

```python
sns.scatterplot(data=df, x='Discount_Used_PCT', y='Total_Orders')
plt.title("Discount % vs Total Orders")
plt.show()
```

### Average Ratings by Category

```python
sns.barplot(data=df, x='Preferred_Category', y='Ratings', estimator='mean')
plt.xticks(rotation=45)
plt.title("Avg Rating by Category")
plt.show()
```

### Relationship Between Variables

```python
sns.pairplot(df[['Age', 'Ratings', 'Total_Orders', 'Discount_Used_PCT']])
plt.show()
```
