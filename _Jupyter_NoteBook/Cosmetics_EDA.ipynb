## Univariate Analysis

### Age Distribution

```python
sns.histplot(df['Age'], kde=True)
plt.title("Age Distribution")
plt.show()
```

### Customer Ratings

```python
sns.countplot(x=df['Ratings'])
plt.title("Ratings Count")
plt.show()
```

### Preferred Category

```python
df['Preferred_Category'].value_counts().plot(kind='bar')
plt.title("Preferred Categories")
plt.show()
```

## Bivariate Analysis

### Age vs Ratings

```python
sns.boxplot(x=df['Ratings'], y=df['Age'])
plt.title("Age vs Ratings")
plt.show()
```

### Category vs Total Orders

```python
sns.barplot(x='Preferred_Category', y='Total_Orders', data=df)
plt.xticks(rotation=45)
plt.title("Orders by Category")
plt.show()
```

### Discount % vs Ratings

```python
sns.scatterplot(x='Discount_Used_PCT', y='Ratings', data=df)
plt.title("Discount vs Ratings")
plt.show()
```
